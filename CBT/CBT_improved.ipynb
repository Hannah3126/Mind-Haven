```python
from llama_cpp import Llama
import os

llm = Llama(
    model_path="/Users/hannahjoshua/Documents/GitHub/Mind-Haven 2/CBT/mistral-7b-instruct-v0.1.Q2_K.gguf",
    n_ctx=512,
    n_threads=4,
    n_gpu_layers=33,  # good for Q2_K on M1
    use_mlock=True,
)
```

```python
def reframe_thought(user_input):
    prompt = f"""### Instruction:
As a compassionate CBT therapist, help reframe the following negative thought into a kind, realistic, and balanced perspective.

### Negative Thought:
{user_input}

### Reframed Thought:"""

    output = llm(
        prompt=prompt,
        max_tokens=80,
        temperature=0.7,
        stop=["###"]  # So it doesn't ramble
    )

    # Extract and return the output text
    return output['choices'][0]['text'].strip()
```

```python
import gradio as gr

interface = gr.Interface(
    fn=reframe_thought,
    inputs=gr.Textbox(label="💭 Your Negative Thought", placeholder="e.g. I failed my test"),
    outputs=gr.Textbox(label="💡 Reframed Thought"),
    title="CBT Thought Reframer (Mistral 7B - Local, Fast)",
    description="Runs fully offline with Mistral on your Mac M1. Fast, secure, and private."
)

interface.launch()
```
